```python
df['AMPLITUDE_PRECO'] = df['PREÇO MÁXIMO REVENDA'] - df['PREÇO MÍNIMO REVENDA']
```

```python
if df['PREÇO MÉDIO REVENDA'] < 5:
    df['CATEGORIA_PRECO'] = "BAIXO"
elif df['PREÇO MÉDIO REVENDA'] > 7:
    df['CATEGORIA_PRECO'] = "ALTO"
elif df['PREÇO MÉDIO REVENDA'] > 5 and df['PREÇO MÉDIO REVENDA'] < 7:
    df['CATEGORIA_PRECO'] = "MEDIO"
```

```python
condicao = [
    df['PREÇO MÉDIO REVENDA'] < 5,
    df['PREÇO MÉDIO REVENDA'] > 7
]

valores = [
    'BAIXO',
    'ALTO'
]

df['CATEGORIA_PRECO'] = np.select(condicao, valores, default = 'MEDIO')
```

```python
df['CATEGORIA_PRECO'].value_counts()
```

```python
def limpar_dados(df):
    df.columns = (df.columns
                 .str.strip()
                 .str.upper()
                 .str.replace(" ", "_")
                 )
    for colunas_numericas in ['PREÇO_MÉDIO_DISTRIBUIÇÃO', 'DESVIO_PADRÃO_DISTRIBUIÇÃO', 'PREÇO_MÍNIMO_DISTRIBUIÇÃO', 'PREÇO_MÁXIMO_DISTRIBUIÇÃO', 'COEF_DE_VARIAÇÃO_DISTRIBUIÇÃO']:
        df[colunas_numericas] = pd.to_numeric(df[colunas_numericas], errors = 'coerce')

    for colunas_data in ['DATA_INICIAL', 'DATA_FINAL']:
        df[colunas_data] = pd.to_datetime(df[colunas_data])
    df = df.fillna('DESCONHECIDO')
    df = df.drop_duplicates()


    return df
```

```python
df = pd.read_csv('./datasets/GasPricesinBrazil_2004-2019.csv', sep=';')
```

```python
limpar_dados(df)
```
