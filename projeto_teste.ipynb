```python
def limpeza_dados(df):
    df.columns = (df.columns
                  .str.strip()
                  .str.upper()
                  .str.replace(' ', '_')
    )
    df['PREÇO_EM_REAIS_POR_MMBTU'] = df['PREÇO_EM_REAIS_POR_MMBTU'].str.replace(',', '.')
    df['PREÇO_EM_REAIS_POR_MMBTU'] = pd.to_numeric(df['PREÇO_EM_REAIS_POR_MMBTU'], errors='coerce')

    df = df.dropna()

    return df
```

```python
limpeza_dados(df)
```

```python
df = df.rename(columns = {
    'TIPOMERCADO' : 'TIPO_MERCADO',
    'REGIÃOAGREGADA': 'REGIÃO_AGREGADA'


})
```

```python
valores_str = ['TIPO_MERCADO', 'REGIÃO_AGREGADA']

for coluna in valores_str:
    df[coluna] = df[coluna].str.upper()
```
