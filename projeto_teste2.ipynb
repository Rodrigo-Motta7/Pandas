```python
def limpeza_dados(df):
    df.columns = (df.columns
                  .str.strip()
                  .str.upper()
                  .str.replace(' ', '_')
    )
                
    colunas_num = ['NOV/21', 'DEZ/21', 'JAN/22', 'FEV/22', 'MAR/22', 'ABR/22','MAI/22', 'JUN/22', 'JUL/22', 'AGO/22', 'SET/22', 'OUT/22' ]
    colunas_str = ['RAZÃO_SOCIAL_(REGISTRO_ANS)', 'COBERTURA', 'PORTE']

    for coluna in colunas_num:
        df[coluna] = df[coluna].str.replace(',', '.')
        df[coluna] = pd.to_numeric(df[coluna], errors='coerce')

    for coluna in colunas_str:
        df[coluna] = df[coluna].str.upper()
    
    df.drop_duplicates()

    return df
    
```

```python
df = limpeza_dados(df)
```

```python
def categorias(classificacao):
    condicao = [
        classificacao < 25,
        classificacao < 50,
        classificacao < 125,
        classificacao >= 125
    ]

    valores = [
        'RUIM',
        'MÉDIO',
        'BOM',
        'MUITO BOM'
    ]

    df['CATEGORIA_CLASSIFICAÇÃO_NO_MÊS'] = np.select(condicao, valores, default='MUITO RUIM')
```

```python
categorias(df['CLASSIFICAÇÃO_NO_MÊS'])
```
