```python
def limpeza_dados(df):
    df.columns = (df.columns
                  .str.strip()
                  .str.upper()
                  .str.replace(' ', '_')
    )

    colunas_str = ['REGIÃO', 'ESTADO', 'PRODUTO']
    colunas_num = ['MARGEM_MÉDIA_REVENDA', 'PREÇO_MÉDIO_DISTRIBUIÇÃO', 'DESVIO_PADRÃO_DISTRIBUIÇÃO', 'PREÇO_MÍNIMO_DISTRIBUIÇÃO', 'PREÇO_MÁXIMO_DISTRIBUIÇÃO', 'COEF_DE_VARIAÇÃO_DISTRIBUIÇÃO']
    for coluna in colunas_str:
        df[coluna] = df[coluna].str.strip().str.upper()

    for coluna in colunas_num:
        df[coluna] = pd.to_numeric(df[coluna], errors = 'coerce')

    for coluna in ['DATA_INICIAL', 'DATA_FINAL']:
        df[coluna] = pd.to_datetime(df[coluna], errors = 'coerce')

    df = df.drop_duplicates()

    return df

    
```

```python
limpeza_dados(df)
```

```python
del df['UNNAMED:_0']
```
