```python
def limpar_dados(df):
    df.columns = (df.columns
        .str.strip()
        .str.upper()
        .str.replace(' ', '_')
    )

    colunas_numericas = ['MARGEM_MÉDIA_REVENDA', 'PREÇO_MÉDIO_DISTRIBUIÇÃO', 'DESVIO_PADRÃO_DISTRIBUIÇÃO', 'PREÇO_MÍNIMO_DISTRIBUIÇÃO', 'PREÇO_MÁXIMO_DISTRIBUIÇÃO', 'COEF_DE_VARIAÇÃO_DISTRIBUIÇÃO']

    for coluna in colunas_numericas:
        df[coluna] = pd.to_numeric(df[coluna], errors = 'coerce')

    for colunas_datas in ['DATA_INICIAL', 'DATA_FINAL']:
        df[colunas_datas] = pd.to_datetime(df[colunas_datas], errors='coerce')

    df = df.dropna(subset=colunas_numericas)

    df= df.drop_duplicates()

    return df

   
       

    
```

```python
def classificar_preco(preco):
    if preco < 5:
        return 'BAIXO'
    elif preco > 7:
        return 'ALTO'
    else:
        return 'MEDIO'
```

```python
df['CATEGORIA_PRECO_2'] = df['PREÇO_MÉDIO_REVENDA'].apply(classificar_preco)
```

```python
df['CATEGORIA_PRECO_2'].unique()
```

```python
def classificar_produto(produto):
    if 'GASOLINA' in produto:
        return "GASOLINA"
    elif 'ETANOL' in produto:
        return "ETANOL"
    elif 'DIESEL' in produto:
        return "DIESEL"
    else:
        return "OUTRO"
```

```python
df['TIPO_COMBUSTIVEL'] = df['PRODUTO'].apply(classificar_produto)
```

```python
df[df['PRODUTO'] == "GLP"]
```

```python
def classificar_preco_detalhado(preco):
    if preco < 5:
        return 'BARATO'
    elif preco >= 5 and preco <= 7:
        return 'NORMAL'
    elif preco > 7 and preco <= 10:
        return 'CARO'
    else:
        return 'MUITO CARO'
```

```python
df['CLASSIFICACAO'] = df['PREÇO_MÉDIO_REVENDA'].apply(classificar_preco_detalhado)
```
