```python
#cria um novo DataFrame
df =pd.DataFrame({
    'A': [1, 2, 3, 4],
    'B': [10, 20, 30, 40],
    'C': [100, 200, 300, 400]  
}, index = ['Linha 1', 'Linha 2', 'Linha 3', 'Linha 4'])
```

```python
#cria uma função para somar a series
def nossa_soma(series):
    return series.sum()
```

```python
#cria uma coluna com a funcao "nossa_soma" que vai somando os valores de cada linha. (axis=1: Linhas/ axis=0: Colunas)
df['SOMA (A, B, C)'] = df.apply(nossa_soma, axis = 1)
```

```python
df
```

```python
#cria um indice com a funcao "nossa_soma" que vai somando os valores de cada coluna. (axis=1: Linhas/ axis=0: Colunas)
df.loc['Linha 5'] = df.apply(nossa_soma, axis = 0)
```

```python
df
```

```python
#cria uma coluna com a função lambda(vc coloca oq vai acontecer dps)que vai pegar a média dos valores de cada linha
#onde so vai pegar os valores das colunas A, B, C
#(axis=1: Linhas/ axis=0: Colunas)
df['Média'] = df[['A', 'B', 'C']].apply(lambda series: series.mean(), axis = 1)
```

```python
df
```

```python
#renomeia o nome da coluna, o inplace serve para a mudança ser no proprio dataframe e n em uma series
df.rename(columns = {
    'Média':'MEDIA(A, B, C)'
}, inplace = True)
```

```python
df
```

```python
#cria uma coluna usando a função lambda onde mostra os valores da coluna C vezes 2.
df['C * 2'] = df['C'].apply(lambda x: x * 2)
```

```python
df
```

```python
#tambem é possivel criar essa coluna de um jeito mais facil, sem precisar usar o lambda
df['A * 2'] = df['A'] * 2
```
